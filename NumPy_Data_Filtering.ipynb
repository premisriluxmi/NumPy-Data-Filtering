# NumPy Data Filtering

## Objective
Boolean indexing is a NumPy technique used to filter array values based on conditions.
In this task, NumPy conditions are used to find values that are above, below, or equal to a given threshold.

```python
import numpy as np

student_names = np.array([
    "Arun", "Dharani", "Kavin", "Premi", "Sowmini",
    "Rohitha", "Keerthana", "Yoga", "Arjun", "Anu"
])

marks = np.array([45, 82, 67, 90, 38, 75, 58, 95, 75, 50])

print("Student Names:", student_names)
print("Student Marks:", marks)
print("Total Number of Students:", len(marks))
```

```python
threshold = 60

above_mask = marks > threshold

print("Threshold:", threshold)
print("Boolean Condition:", above_mask)
print("Marks above", threshold, ":", marks[above_mask])
print("Number of students:", len(marks[above_mask]))

print("\nStudents scoring above", threshold, ":")
for name, mark in zip(student_names[above_mask], marks[above_mask]):
    print(name, "-", mark)
```

```python
threshold = 50

below_mask = marks < threshold

print("Threshold:", threshold)
print("Boolean Condition:", below_mask)
print("Marks below", threshold, ":", marks[below_mask])
print("Number of students:", len(marks[below_mask]))

print("\nStudents scoring below", threshold, ":")
for name, mark in zip(student_names[below_mask], marks[below_mask]):
    print(name, "-", mark)
```

```python
threshold = 75

equal_mask = marks == threshold

print("Threshold:", threshold)
print("Boolean Condition:", equal_mask)
print("Marks equal to", threshold, ":", marks[equal_mask])
print("Number of students:", len(marks[equal_mask]))

print("\nStudents scoring exactly", threshold, ":")
for name, mark in zip(student_names[equal_mask], marks[equal_mask]):
    print(name, "-", mark)
```

```python
thresholds = [50, 60, 75, 90]

for t in thresholds:
    print("-" * 30)
    print("Threshold:", t)
    print("Above :", marks[marks > t])
    print("Below :", marks[marks < t])
    print("Equal :", marks[marks == t])

print("-" * 30)
```
